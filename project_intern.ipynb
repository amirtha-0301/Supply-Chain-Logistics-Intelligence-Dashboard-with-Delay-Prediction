```python
df_clean = df.copy()

print("Backup Created Successfully!")
```

```python
missing = df_clean.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

missing
```

```python


print("Duplicate Rows Before:", df_clean.duplicated().sum())

df_clean.drop_duplicates(inplace=True)

print("Duplicate Rows After :", df_clean.duplicated().sum())
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("/", "_")
)

print(df_clean.columns.tolist())
```

```python
df_clean["order_date_dateorders"] = pd.to_datetime(
    df_clean["order_date_dateorders"], errors="coerce"
)

df_clean["shipping_date_dateorders"] = pd.to_datetime(
    df_clean["shipping_date_dateorders"], errors="coerce"
)

print("Date Conversion Completed!")
```

```python
df_clean["delivery_days"] = (
    df_clean["shipping_date_dateorders"] -
    df_clean["order_date_dateorders"]
).dt.days
df_clean["order_year"] = df_clean["order_date_dateorders"].dt.year
df_clean["order_month"] = df_clean["order_date_dateorders"].dt.month_name()
df_clean["order_day"] = df_clean["order_date_dateorders"].dt.day_name()
print("New Features Created Successfully!")
```

```python
conn = sqlite3.connect("supply_chain.db")

print("Database Created Successfully!")
```

```python
df_clean.to_sql(
    name="supply_chain",
    con=conn,
    if_exists="replace",
    index=False
)

print("Dataset Exported to SQL Successfully!")
```

```python
query = """
SELECT *
FROM supply_chain
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT COUNT(*) AS Total_Orders
FROM supply_chain;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT ROUND(SUM(sales),2) AS Total_Sales
FROM supply_chain;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT customer_segment,
       ROUND(SUM(sales),2) AS Total_Sales
FROM supply_chain
GROUP BY customer_segment
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT category_name,
       ROUND(SUM(sales),2) AS Total_Sales
FROM supply_chain
GROUP BY category_name
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT market,
       ROUND(SUM(sales),2) AS Total_Sales
FROM supply_chain
GROUP BY market
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT ROUND(AVG(delivery_days),2) AS Average_Delivery_Days
FROM supply_chain;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT shipping_mode,
       COUNT(*) AS Total_Orders
FROM supply_chain
GROUP BY shipping_mode
ORDER BY Total_Orders DESC;
"""

pd.read_sql(query, conn)
```

```python

query = """
SELECT customer_fname || ' ' || customer_lname AS Customer_Name,
       ROUND(SUM(sales),2) AS Total_Sales
FROM supply_chain
GROUP BY Customer_Name
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```
